# EX:12 Write a Python Program to handle missing data using pandas

```python
import pandas as pd
import numpy as np

# 3. Create a DataFrame containing missing values
data = {
    "Name": ["Arun", "Bala", "Cathy", "David", "Eva"],
    "Marks": [85, np.nan, 78, np.nan, 90],
    "Age": [18, 19, np.nan, 20, np.nan]
}

df = pd.DataFrame(data)

# 4. Display the original DataFrame
print("Original DataFrame:")
print(df)

# 5. Check and display missing values
print("\nMissing Values:")
print(df.isnull())

# 6. Count missing values in each column
print("\nNumber of Missing Values in Each Column:")
print(df.isnull().sum())

# 7. Fill all missing values with a default value
df_default = df.fillna(0)

print("\nDataFrame after filling missing values with 0:")
print(df_default)

# 8. Fill missing values in Marks column with column mean
df_mean = df.copy()
mean_marks = df_mean["Marks"].mean()
df_mean["Marks"] = df_mean["Marks"].fillna(mean_marks)

print("\nMarks column after filling missing values with mean:")
print(df_mean)

# 9. Perform forward fill
df_ffill = df.ffill()

print("\nDataFrame after Forward Fill:")
print(df_ffill)

# 10. Perform backward fill
df_bfill = df.bfill()

print("\nDataFrame after Backward Fill:")
print(df_bfill)

# 11. Remove rows containing missing values
df_drop = df.dropna()

print("\nDataFrame after removing rows with missing values:")
print(df_drop)

# 12. Replace a specified value
df_replace = df.fillna(0)
df_replace = df_replace.replace(0, 100)

print("\nDataFrame after replacing 0 with 100:")
print(df_replace)

# 13. Display DataFrame information
print("\nDataFrame Information:")
df.info()
```

# EX:13 Write a python program to group and aggregate data in pandas

```python
# Grouping and Aggregation Data using pandas

import pandas as pd

# Creating DataFrame
data = {
    "Department": ["HR", "IT", "HR", "IT", "Sales", "Sales"],
    "Employee": ["Amit", "Rahul", "Priya", "Ankit", "Sakshi", "Miya"],
    "Salary": [30000, 50000, 35000, 60000, 45000, 40000],
    "Experience": [2, 5, 3, 6, 4, 2]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Grouping by Department
grouped = df.groupby("Department")

# Sum Aggregation
print("\nSum of Salary by Department:")
print(grouped["Salary"].sum())

# Mean Aggregation
print("\nAverage Salary by Department:")
print(grouped["Salary"].mean())

# Maximum Salary
print("\nMaximum Salary by Department:")
print(grouped["Salary"].max())

# Minimum Salary
print("\nMinimum Salary by Department:")
print(grouped["Salary"].min())

# Multiple Aggregations
print("\nMultiple Aggregations:")
print(
    grouped["Salary"].agg(
        ["sum", "mean", "max", "min"]
    )
)

# Grouping Multiple Columns
print("\nGrouping by Department and Experience:")

group_multi = df.groupby(
    ["Department", "Experience"]
)["Salary"].sum()

print(group_multi)

# Iterating Groups
print("\nDisplaying Each Group:")

for name, group in grouped:
    print("\nDepartment:", name)
    print(group)

# Sorting Values
print("\nSorted DataFrame by Salary:")

sorted_df = df.sort_values(by="Salary")
print(sorted_df)
```

# EX:14 Write a Python Program to merge and join DataFrames using Pandas

```python
# Merge and Join DataFrames using Pandas

import pandas as pd

# ----- CONCATENATE OPERATIONS ------------------- (all columns name should be same)

# Creating First DataFrame
df1 = pd.DataFrame({
    "ID": [101, 102, 103, 105],
    "Marks": [95, 88, 90, 85]
}, index=[0, 1, 2, 3])   

print("\nDataFrame 1:")
print(df1)

#  Creating Second DataFrame
df2 = pd.DataFrame({
    "ID": [201, 202, 203, 205],
    "Marks": [85, 78, 80, 65]
}, index=[4, 5, 6, 7])   

print("\nDataFrame 2:")
print(df2)

frames = [df1,df2]
result = pd.concat(frames)
print("concatenate dataframes")
print(result)

# ---------------- MERGE OPERATIONS ---------------- ( at least one column should be same, others can be different  )

# Creating First DataFrame
df1 = pd.DataFrame({
    "ID": [101, 102, 103, 104],
    "Name": ["Sakshi", "Rahul", "Ankit", "Priya"]
})

print("DataFrame 1:")
print(df1)

# Creating Second DataFrame
df2 = pd.DataFrame({
    "ID": [101, 102, 103, 105],
    "Marks": [95, 88, 90, 85]
}, index=[0, 1, 2, 3])   

print("\nDataFrame 2:")
print(df2)

# Inner Merge
inner_merge = pd.merge(df1, df2, on="ID", how="inner")

print("\nInner Merge:")
print(inner_merge)

# Left Merge
left_merge = pd.merge(df1, df2, on="ID", how="left")

print("\nLeft Merge:")
print(left_merge)

# Right Merge
right_merge = pd.merge(df1, df2, on="ID", how="right")

print("\nRight Merge:")
print(right_merge)

# Outer Merge
outer_merge = pd.merge(df1, df2, on="ID", how="outer")

print("\nOuter Merge:")
print(outer_merge)

# ---------------- JOIN OPERATIONS ---------------- ( columns names in both datframes is different )

# Creating DataFrames for Join
df3 = pd.DataFrame({
    "Name": ["Amit", "Riya", "Karan"]
}, index=[1, 2, 3])

df4 = pd.DataFrame({
    "Salary": [30000, 40000, 50000]
}, index=[1, 2, 3])

print("\nDataFrame 3:")
print(df3)

print("\nDataFrame 4:")
print(df4)

# Joining DataFrames
joined_df = df3.join(df4)

print("\nJoined DataFrame:")
print(joined_df)

# Joining with Different Join Types
left_join = df3.join(df4, how="left")

print("\nLeft Join:")
print(left_join)

right_join = df3.join(df4, how="right")

print("\nRight Join:")
print(right_join)

outer_join = df3.join(df4, how="outer")

print("\nOuter Join:")
print(outer_join)
```
